```python
from __future__ import unicode_literals
from os import replace
a = input('문장을 입력')
b = input('문장을 입력')
c = input('문장을 입력')
d =a+''+b+''+c
print(d.replace(',',''))
print(d.lower())
words = d.split()
unicode_literals = set(words)
sorted_unique_words = sorted(unicode_literals )
print('\n단어 목록')
print(sorted_unique_words)
```

```python
import random
mem = ['김민준', ' 이서준', '박도윤', '최시우','정예준', '강주원','조유준']
winners_cnt = int(input('뽑을 당첨자 수를 입력력하세요:'))
winers = random.sample(mem, winners_cnt)
winers.sort()
print(winers)
```
