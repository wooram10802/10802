```python
def a(b, c, d):
  if c == '+':
    return b+d
  elif c == '-':
    return b-d
  elif c == '*':
    return  b*d
  elif c == '/':
    if d == 0:
      return "오류"
    return b / d
  else:
    return('오류')
print(a(10,'+',5))
print(a(10,'-',5))
print(a(10,'*',5))
print(a(10,'/',5))
```
