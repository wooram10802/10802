```python
from gtts import gTTS
text = "안녕하세요. gTTS를 이용해 텍스트를 음성으로 변환하고 있습니니다."
tts = gTTS(text=text, lang="ko")
tts.save("output.mp3")
print("음성 파일 저장 완료!")
```
